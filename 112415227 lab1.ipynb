Extend the function to take another optional flag parameter and add only if it is true






```python
def add(x, y, z=None, flag=False):
    if flag:
        if z is not None:
            return x + y + z
        return x + y
    else:
        if z is not None:
            return x + y + z
        return x + y

# Simplified clean version:
def add(x, y, z=None, flag=False):
    total = x + y
    if z is not None:
        total += z
    if flag:
        return total
    return total

print(add(1, 2))
print(add(1, 2, 3))
print(add(1, 2, 3, flag=True))
```

Show how to use a iterator. First, use enumerate to create a new tuple from above list

```python
#use of a iterator
#iterator- returns numbers that starts with 25 and each and every time it is passed it gets squared.

class MyInit:
    def __init__(self):
        self.a = 25

    def __iter__(self):
        return self

    def __next__(self):
        x = self.a
        self.a = self.a ** 2
        return x

obj1 = MyInit()
itr = iter(obj1)

print(next(itr))  # Returns 25
print(next(itr))  # Returns 625
```

## Strings in Python

Open the file poem.txt and print lenght of each line.

```python
import statistics

line_lengths = []
all_words = []
full_text = ""

with open('poem.txt', 'r') as file:
    for line in file:
        line_lengths.append(len(line))
        all_words.extend(line.split())
        full_text += line

# Calculated metrics
total_lines_len = sum(line_lengths)
mean_len = statistics.mean(line_lengths) if line_lengths else 0
median_len = statistics.median(line_lengths) if line_lengths else 0
std_dev_len = statistics.stdev(line_lengths) if len(line_lengths) > 1 else 0

print(f"Lines Processed: {len(line_lengths)}")
```

Find sum of each line, mean, median and standard deviation of all lines in the file.

```python
print(f"Sum: {total_lines_len}, Mean: {mean_len:.2f}, Median: {median_len}, StdDev: {std_dev_len:.2f}")
```

# Working with CSV

Let's import a file with fuel economy data

```python
import csv

with open('mpg.csv') as csvfile:
    mpg = list(csv.DictReader(csvfile))
```

Find out how many items are there in the dictionary

```python
print("Number of columns/items per row:", len(mpg[0].items()))
```

Find the column names with keys

```python
print("Column names:", list(mpg[0].keys()))
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```python
avg_cty = sum(float(d['cty']) for d in mpg) / len(mpg)
print(f"Average City MPG: {avg_cty:.2f}")
```

Find the average hwy fuel economy across all cars

```python
avg_hwy = sum(float(d['hwy']) for d in mpg) / len(mpg)
print(f"Average Highway MPG: {avg_hwy:.2f}")
```

Use set to return the unique values for the number of cylinders the cars.

```python
cylinders = set(d['cyl'] for d in mpg)
print("Cylinders:", cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```python
MpgByCyl = []

for c in cylinders:
    summpg = 0
    typecount = 0
    for d in mpg:
        if d['cyl'] == c:
            summpg += float(d['cty'])
            typecount += 1
    MpgByCyl.append((c, summpg / typecount))

MpgByCyl.sort(key=lambda x: x[0])
print(MpgByCyl)
```

Return the unique values for the class types in our dataset using set.

```python
vehicle_classes = set(d['class'] for d in mpg)
print(vehicle_classes)
```

Find average miles per gallon in the highway for each class of vehicles

```python
vehicleclass = set(d['class'] for d in mpg)

MileageByClass = []

for t in vehicleclass:
    sumpg = 0
    classCount = 0
    for d in mpg:
        if d['class'] == t:
            sumpg += float(d['hwy'])
            classCount += 1
    MileageByClass.append((t, sumpg / classCount))

MileageByClass.sort(key=lambda x: x[1])
print(MileageByClass)
```
